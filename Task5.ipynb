```python
import matplotlib.pyplot as plt
```

```python
# Student study analysis

days = ['Monday', 'Tuesday', 'Wednesday', 'Thursday', 'Friday', 'Saturday', 'Sunday']

python = [2, 3, 2, 4, 3, 2, 4]

sql = [1, 1, 2, 1, 2, 1, 1]

mathematics = [1, 2, 1, 1, 2, 2, 1]

revision = [0.5, 1, 0.5, 1, 1, 2, 2]

plt.figure(figsize=(10 , 5))

plt.stackplot(
    days,
    python,
    sql,
    mathematics,
    revision,
    labels=[
        "Python",
        "SQL",
        "Mathematics",
        "Revision",
    ],
    colors=[
        "red",
        "black",
        "yellow",
        "blue"
    ]
)

plt.title("Students study analysis")

plt.xlabel("Days")

plt.ylabel("Hours")

plt.legend()

plt.show()
```
