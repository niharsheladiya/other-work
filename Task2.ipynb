```python
import matplotlib.pyplot as plt
```

```python
# Monthly Product Sales Analysis

months = ['JAN', 'FEB', 'MAR', 'APR', 'MAY', 'JUN']

electronics = [12000, 13500, 12800, 14200, 15000, 16500]

clothing = [8500, 9200, 10500, 11000, 12500, 13000]

grocery = [5000, 5200, 5100, 5400, 5800, 6000]

plt.figure(figsize=(10 , 5))

plt.stackplot(
    months,
    electronics,
    clothing,
    grocery,
    labels=[
        "Electronics",
        "Clothing",
        "Grocery",
    ],
    colors=[
        "red",
        "blue",
        "Green"
    ]
)

plt.title("Monthly Product Sales Analysis")

plt.xlabel("Month")

plt.ylabel("Sales")

plt.legend()

plt.show()
```
