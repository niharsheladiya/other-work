```python
import matplotlib.pyplot as plt
```

```python
# Restaurant Sales

days = ['Monday', 'Tuesday', 'Wednesday', 'Thursday', 'Friday', 'Saturday', 'Sunday']

breakfast = [300, 320, 310, 340, 400, 500, 550]

lunch = [600, 650, 620, 680, 800, 950, 1000]

dinner = [800, 850, 800, 900, 1200, 1500, 1400]

plt.figure(figsize=(10 , 5))

plt.stackplot(
    days,
    breakfast,
    lunch,
    dinner,
    labels=[
        "Breakfast",
        "Lunch",
        "Dinner",
    ],
    colors=[
        "lightgreen",
        "lightblue",
        "lightcoral",
    ]
)

plt.title("Restaurant Sales")

plt.xlabel("Days")

plt.ylabel("Sales Amount")

plt.legend()

plt.show()
```
