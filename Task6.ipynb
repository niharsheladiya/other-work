```python
import matplotlib.pyplot as plt
```

```python
# E-Commrce Orders

months = ['JAN', 'FEB', 'MAR', 'APR', 'MAY', 'JUN']

electronics = [1500, 1600, 1750, 1900, 2100, 2300]

fashion = [2500, 2700, 2650, 2900, 3100, 3300]

grocery = [3200, 3100, 3400, 3550, 3800, 4000]

plt.figure(figsize=(10 , 5))

plt.stackplot(
    months,
    electronics,
    fashion,
    grocery,
    labels=[
        "Electronics",
        "Fashion",
        "Grocery",
    ],
    colors=[
        "pink",
        "orange",
        "yellow"
    ]
)

plt.title("E-Commerce Orders")

plt.xlabel("Months")

plt.ylabel("Numbers of orders")

plt.legend()

plt.show()
```
