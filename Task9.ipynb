```python
import matplotlib.pyplot as plt
```

```python
# Fitness Activity

days = ['Monday', 'Tuesday', 'Wednesday', 'Thursday', 'Friday', 'Saturday', 'Sunday']

walking = [30, 45, 40, 50, 45, 60, 75]

running = [15, 20, 0, 25, 20, 40, 35]

cycling = [0, 30, 45, 20, 0, 60, 90]

plt.figure(figsize=(10 , 5))

plt.stackplot(
    days,
    walking,
    running,
    cycling,
    labels=[
        "Walking",
        "Running",
        "Cycling",
    ],
    colors=[
        "pink",
        "lightblue",
        "orange"
    ]
)

plt.title("Fitnees Activity")

plt.xlabel("Days")

plt.ylabel("Minutes")

plt.legend()

plt.show()
```
