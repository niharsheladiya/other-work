```python
import matplotlib.pyplot as plt
```

```python
# Monthly visitors of websites

months = ['JAN' , 'FEB' , 'MAR' , 'APR' , 'MAY' , 'JUN']

google_search = [1500, 1650, 1800, 1950, 2100, 2400]

social_media = [800, 950, 1100, 1050, 1250, 1400]

direct_visitors = [400, 450, 480, 520, 580, 650]

plt.figure(figsize=(8 , 5))

plt.stackplot(
  months,
  google_search,
  social_media,
  direct_visitors,
  labels=[
    "Google Search",
    "Social Media",
    "Direct Visitors"
  ],
  colors=[
    "skyblue",
    "orange",
    "lightgreen"
  ]
)

plt.title("Monthly Websites Visitors")

plt.xlabel("Month")

plt.ylabel("Visitors")

plt.legend()

plt.show()
```
