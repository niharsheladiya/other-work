```python
import matplotlib.pyplot as plt
```

```python
#  Employee works hours

days = ['Monday', 'Tuesday', 'Wednesday', 'Thursday', 'Friday']

coding = [4, 5, 3, 6, 4]

meetings = [2, 1, 3, 1, 2]

training = [2, 2, 2, 1, 2]

plt.figure(figsize=(10 , 5))

plt.stackplot(
    days,
    coding,
    meetings,
    training,
    labels=[
        "Coding",
        "Meetings",
        "Training",
    ],
    colors=[
        "black",
        "green",
        "blue"
    ]
)

plt.title("Employee Works hour")

plt.xlabel("Days of the week")

plt.ylabel("Hours")

plt.legend()

plt.show()

```
