```python
import matplotlib.pyplot as plt
```

```python
# Mobile app usage

months = ['JAN', 'FEB', 'MAR', 'APR', 'MAY', 'JUN']

videos = [40, 45, 42, 50, 55, 60]

games = [30, 32, 35, 38, 40, 45]

chat = [20, 22, 25, 24, 28, 30]

shopping = [10, 12, 15, 18, 20, 25]

plt.figure(figsize=(10 , 5))

plt.stackplot(
    months,
    videos,
    games,
    chat,
    shopping,
    labels=[
        "Videos",
        "Games",
        "Chat",
        "Shopping",
    ],
    colors=[
        "purple",
        "yellow",
        "skyblue",
        "orange"
    ]
)

plt.title("Mobile App Usage")

plt.xlabel("Months")

plt.ylabel("Usage Hour")

plt.legend()

plt.show()
```
