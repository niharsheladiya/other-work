#### 2. Combining and Splitting Arraya

```python
print("Horizontal and Vertical Concatination")

arr1 = np.array([1 , 2 , 3])
arr2 = np.array([4 , 5 , 6])

print(arr1)
print(arr2)

horizontal = np.hstack((arr1, arr2))

vertical = np.vstack((arr1, arr2))

print(horizontal)
print(vertical)
```
