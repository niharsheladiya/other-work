```python
import matplotlib.pyplot as plt
```

```python
# YouTube Channel Performance

months = ['JAN', 'FEB', 'MAR', 'APR', 'MAY', 'JUN']

tutorials = [5000, 5200, 5500, 5800, 6000, 6500]

shorts = [10000, 15000, 18000, 20000, 25000, 30000]

reviews = [2000, 2100, 2300, 2500, 2400, 2800]

plt.figure(figsize=(10 , 5))

plt.stackplot(
    months,
    tutorials,
    shorts,
    reviews,
    labels=[
        "Tutorials",
        "Shorts",
        "Reviews",
    ],
    colors=[
        "yellow",
        "green",
        "blue"
    ]
)

plt.title("Youtube Channel Performance")

plt.xlabel("Months")

plt.ylabel("Views")

plt.legend()

plt.show()
```
