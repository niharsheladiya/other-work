```python
import matplotlib.pyplot as plt
```

```python
# Electricity Consumption

months = ['JAN', 'FEB', 'MAR', 'APR', 'MAY', 'JUN']

ac = [120, 150, 180, 210, 200, 160]

refrigerator = [50, 52, 55, 55, 54, 50]

lights = [30, 28, 25, 25, 28, 32]

other_appliances = [45, 50, 48, 55, 42, 40]

plt.figure(figsize=(10 , 5))

plt.stackplot(
    months,
    ac,
    refrigerator,
    lights,
    other_appliances,
    labels=[
        "AC",
        "Refrigerator",
        "Lights",
        "Other_Appliances",
    ],
    colors=[
        "lightgreen",
        "pink",
        "lightcoral",
        "lightblue"
    ]
)

plt.title("Elecricity Consumption")

plt.xlabel("Months")

plt.ylabel("Electricity Units")

plt.legend()

plt.show()
```
